```python
import yfinance as yf
import pandas as pd

tickers = [
'^GSPC', 'VBMFX'
]
start_date = "2000-01-31"
end_date = "2002-12-31"

# Download data
prices = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=False
)["Adj Close"]

# Ensure it's a DataFrame even if only one ticker
if isinstance(prices, pd.Series):
    prices = prices.to_frame()

# Name the index
prices.index.name = "Date"

# Export to Excel
prices.to_excel("adjusted_close_prices.xlsx")

print(prices.head())
    
```
